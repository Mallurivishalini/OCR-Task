```python
import easyocr
from PIL import Image
import requests
```

```python
reader = easyocr.Reader(['en'], gpu=False)
```

```python
def extract_text_from_image(image_path):
    """
    Extract text from an image using EasyOCR.
    """
    results = reader.readtext(image_path, detail=0)  # Only text, no bounding boxes
    return results
```

```python
def search_openfoodfacts(query):
    """
    Search for product in Open Food Facts API using text query.
    """
    api_url = "https://world.openfoodfacts.org/cgi/search.pl"
    params = {
        'search_terms': query,
        'search_simple': 1,
        'action': 'process',
        'json': 1
    }
    response = requests.get(api_url, params=params)
    if response.status_code == 200:
        data = response.json()
        if data['count'] > 0:
            product = data['products'][0]
            return {
                'Product Name': product.get('product_name', 'N/A'),
                'Brand': product.get('brands', 'N/A'),
                'Ingredients': product.get('ingredients_text', 'N/A'),
                'Nutrition Facts': product.get('nutriments', {}),
                'Categories': product.get('categories', 'N/A')
            }
        else:
            return {"Error": "No product found"}
    else:
        return {"Error": "API request failed"}
```

```python
extracted_texts = extract_text_from_image(image_path)
print("Extracted Text:", extracted_texts)
```

```python
product_info = None
for text in extracted_texts:
    result = search_openfoodfacts(text)
    if 'Error' not in result:
        product_info = result
        break
```
