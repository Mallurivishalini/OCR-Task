# OCR-Based Product Information Scanner

## 📌 Overview

This project is an **OCR-powered product label scanner** that:
1. Extracts text (product name, barcode, etc.) from an image using EasyOCR.
2. Queries the **Open Food Facts API** to fetch product details.
3. Displays key information like product name, brand, ingredients, nutrition facts, and categories in a user-friendly interface.

The application can be run in a **Jupyter Notebook** or via a simple **Streamlit web interface**.

## ⚙️ Features

- OCR extraction using EasyOCR
- Barcode or product name search using Open Food Facts API
- CLI and Streamlit-based output options
- Handles missing data and fallback search methods

## 🖥️ Tech Stack

- **Python 3.10.8**
- EasyOCR
- PyTorch
- OpenCV
- Requests
- Streamlit 

## 📦 Setup Instructions

### 1️⃣ Create and activate environment

 **Note:** PyTorch and EasyOCR have better support for Python 3.10 or 3.11.

conda create -n ocr_env python=3.10 -y
conda activate ocr_env


### 2️⃣ Install dependencies

pip install torch torchvision easyocr requests opencv-python matplotlib pillow streamlit

### 3️⃣ Add environment to Jupyter

python -m ipykernel install --user --name=ocr_env --display-name "Python (ocr_env)"

## 🚀 Usage

### In Jupyter Notebook
1.Open the notebook ocr_product_scanner.ipynb

2.Make sure your kernel is set to Python (ocr_env)

3.Run the cells to:
>Extract text from a sample image

>Search the Open Food Facts database

>Display product details

## In Streamlit
1.Save the Streamlit code from the notebook into app.py

2.Run: "streamlit run app.py"

3.Upload an image to see OCR results and product info in your browser.

## 🛠 Approach

#### OCR Extraction

Use EasyOCR to read text from the uploaded image.

Detect both product names and barcodes (numeric patterns).

#### Data Retrieval

If a barcode is detected, query the Open Food Facts API by barcode.

Otherwise, use the product name for search.

API returns JSON containing brand, categories, nutrition facts, etc.

#### Display Results

Present extracted text and matched product data in either CLI or web UI.

Handle missing fields gracefully.
